```python
import yfinance as yf
import pandas as pd
import numpy as np
import time
from datetime import datetime

# 1. Carregar a lista de tickers
try:
    df_inicial = pd.read_csv('lista_tickers.csv')
    # Tenta encontrar a coluna 'ticker' independente de maiúsculas/minúsculas
    col_ticker = next((c for c in df_inicial.columns if c.upper() == 'TICKER'), None)

    if col_ticker:
        tickers = [str(t).strip() + ".SA" if not str(t).strip().endswith(".SA") else str(t).strip() for t in df_inicial[col_ticker].tolist()]
    else:
        print(f"Erro: Coluna 'TICKER' não encontrada no arquivo. Colunas disponíveis: {df_inicial.columns.tolist()}")
        tickers = []
except FileNotFoundError:
    print("Erro: Arquivo 'lista_tickers.csv' não encontrado.")
    tickers = []

dados_finais = []
ano_atual = datetime.now().year

print(f"Iniciando coleta de dados para {len(tickers)} ativos...")

for symbol in tickers:
    try:
        ticker_obj = yf.Ticker(symbol)
        info = ticker_obj.info

        # Preço e Volume para Liquidez
        preco = info.get('currentPrice') or info.get('regularMarketPreviousClose') or 0
        vol_medio_3m = info.get('averageVolume', 0)
        liquidez_media_3m = vol_medio_3m * preco

        # Payout
        payout = info.get('payoutRatio', 0)

        # Lucro Líquido
        lucro_liquido = info.get('netIncomeToCommon', 0)

        # LPA dos últimos 20 trimestres (5 anos)
        try:
            q_financials = ticker_obj.quarterly_financials
            if 'Basic EPS' in q_financials.index:
                lpa_hist = q_financials.loc['Basic EPS'].head(20).tolist()
            else:
                lpa_hist = []
        except:
            lpa_hist = []

        # Dividend Yield dos últimos 3 anos
        divs = ticker_obj.dividends
        dy_data = {}
        for i in range(1, 4):
            ano = ano_atual - i
            soma_div = divs[divs.index.year == ano].sum() if not divs.empty else 0
            dy_data[f'DY_{ano}'] = (soma_div / preco) if preco > 0 else 0

        # Verificação de Recuperação Judicial
        resumo = str(info.get('longBusinessSummary', '')).lower()
        em_recuperacao = 1 if 'recuperação judicial' in resumo or 'judicial recovery' in resumo else 0

        dados_finais.append({
            'Ticker': symbol.replace(".SA", ""),
            'Empresa': info.get('longName', ''),
            'Liquidez_Media_3M': liquidez_media_3m,
            'Payout': payout,
            'Lucro_Liquido_Atual': lucro_liquido,
            'DY_Ano_1': dy_data.get(f'DY_{ano_atual-1}', 0),
            'DY_Ano_2': dy_data.get(f'DY_{ano_atual-2}', 0),
            'DY_Ano_3': dy_data.get(f'DY_{ano_atual-3}', 0),
            'Recuperacao_Judicial': em_recuperacao,
            'LPA_Trimestral_Hist': " | ".join(map(str, lpa_hist))
        })
        print(f"Sucesso: {symbol}")
        time.sleep(0.1)

    except Exception as e:
        print(f"Erro em {symbol}: {e}")

# Criar DataFrame e Salvar
if dados_finais:
    df_resultado = pd.DataFrame(dados_finais)
    df_resultado.to_csv('base_lovable_barsi.csv', index=False)
    print(f"\nConcluído! Arquivo 'base_lovable_barsi.csv' gerado com {len(df_resultado)} registros.")
    display(df_resultado.head())
else:
    print("Nenhum dado foi coletado.")
```
